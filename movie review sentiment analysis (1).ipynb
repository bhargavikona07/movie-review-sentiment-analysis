```python
import pandas as pd
df=pd.read_csv("IMDB Dataset.csv")
review=input("Enter movie review: ").lower()
positive_words=["good", "great", "excellent", "amazing","awesome", "wonderful", "fantastic", "best", "love"]
negative_words=["bad", "poor", "boring", "terrible","worst", "waste", "hate", "awful"]
positive_count=0
negative_count=0
for word in positive_words:
    if word in review:
        positive_count+=1
for word in negative_words:
    if word in review:
        negative_count+=1
if positive_count>negative_count:
    print("Prediction: POSITIVE")
elif negative_count>positive_count:
    print("Prediction: NEGATIVE")
else:
    print("Prediction: NEUTRAL")
```
